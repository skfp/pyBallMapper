```python
import numpy as np
from sklearn.datasets import make_blobs, make_friedman1
from sklearn.preprocessing import StandardScaler
# from tqdm import tqdm
import time
from matplotlib import pyplot as plt
from pyballmapper import BallMapper
```

# Compare random epsilon net with distance-based algorithm

```python
# Create a sample dataset
n = 1000
X, _ = make_blobs(n_samples=n, centers=((-10,0),(-2,2),(5,-1)), random_state=42)
plt.scatter(X[:,0],X[:,1])
```

```python
# Create Ball Mapper graph using nearest neighbor landmark selection

bm = BallMapper(
    X=X,  # the pointcloud, as a numpy array
    eps=2, # the radius of the balls
    method="nearest",  # use the nearest uncovered point based landmark selection
)  

bm.draw_networkx()
```
